Visualização de dados usando Matplotlib

```python
# Importando bibliotecas
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Carregando a base de dados
df_district = pd.read_csv('/content/drive/MyDrive/Tópicos Especiais - Engenharia de Dados/Projeto 1/district.csv', sep=',')
df_cts = pd.read_csv('/content/drive/MyDrive/Tópicos Especiais - Engenharia de Dados/Projeto 1/case_time_series.csv', sep=',')
```

```python
# Mostrando como é o dataset 2
df_cts.head()
```

Obs.: Dos dois datasets usados, apenas o dataset 2 (case_times_series.csv) possui os dados organizados no tempo. Já o dataset 1 (district.csv) nos indica apenas o número de casos de acordo com a cidade.

Portanto, iremos usar o dataset 2 para apresentar os gráficos pedidos nesse projeto.

```python
# Obtendo as informações do dataset 2
df_cts.describe()
```

```python
### Análise do números de casos diários

# Vamos remodelar o dataset 2 para ter apenas as colunas com dados diários
df_daily = df_cts[['Date','Daily Confirmed','Daily Recovered','Daily Deceased']]
df_daily.head()
```

```python
### Análise do número de casos acumulados

# E, teremos outro dataset com os casos acumulados de acordo com o tempo
df_t = df_cts[['Date','Total Confirmed','Total Recovered','Total Deceased']]
df_t.head()
```

```python
# Gráficos em linhas

# Casos diários
plt.plot(df_daily['Date'], df_daily['Daily Confirmed'], label='Confirmados')
plt.plot(df_daily['Date'], df_daily['Daily Recovered'], label='Recuperados')
plt.plot(df_daily['Date'], df_daily['Daily Deceased'], label='Mortes')
plt.legend()
plt.xticks([])
plt.xlabel('Tempo em Dias (30/01 a 13/05)')
plt.ylabel('Número de Casos')
plt.title('Casos Diários de COVID-19 na Índia')

# Gráfico obtido
plt.show()
```

```python
# Gráficos em linhas

# Casos acumulados
plt.plot(df_t['Date'], df_t['Total Confirmed'], label='Confirmados')
plt.plot(df_t['Date'], df_t['Total Recovered'], label='Recuperados')
plt.plot(df_t['Date'], df_t['Total Deceased'], label='Mortes')
plt.legend()
plt.xticks([])
plt.xlabel('Tempo em Dias (30/01 a 13/05)')
plt.ylabel('Número de Casos Acumulados')
plt.title('Casos Acumulados de COVID-19 na Índia')

# Gráfico obtido
plt.show()
```

```python
# Gráfico em barras

# Casos diários
plt.bar(df_daily['Date'], df_daily['Daily Confirmed'], label="Confirmados", color='r')
plt.bar(df_daily['Date'], df_daily['Daily Recovered'], label="Recuperados", color='g')
plt.bar(df_daily['Date'], df_daily['Daily Deceased'], label='Mortes', color='b')
plt.legend()
plt.xticks([])
plt.xlabel('Tempo em Dias (30/01 a 13/05)')
plt.ylabel('Número de Casos')
plt.title('Casos Diários de COVID-19 na Índia')

# Gráfico obtido
plt.show()
```

```python
# Gráfico em barras

# Casos acumulados
plt.bar(df_t['Date'], df_t['Total Confirmed'], label="Confirmados", color='m')
plt.bar(df_t['Date'], df_t['Total Recovered'], label="Recuperados", color='y')
plt.bar(df_t['Date'], df_t['Total Deceased'], label='Mortes', color='g')
plt.legend()
plt.xticks([])
plt.xlabel('Tempo em Dias (30/01 a 13/05)')
plt.ylabel('Número de Casos Acumulados')
plt.title('Casos Acumulados de COVID-19 na Índia')

# Gráfico obtido
plt.show()
```

```python
# Gráfico de dispersão

# Casos diários
plt.scatter(df_daily['Date'], df_daily['Daily Confirmed'], label='Confirmados', color='m', marker='o', s=100)
plt.scatter(df_daily['Date'], df_daily['Daily Recovered'], label='Recuperados', color='b', marker='o', s=100)
plt.scatter(df_daily['Date'], df_daily['Daily Deceased'], label='Mortos', color='c', marker='+', s=100)
plt.legend()
plt.xticks([])
plt.xlabel('Tempo em Dias (30/01 a 13/05)')
plt.ylabel('Número de Casos')
plt.title('Casos Diários de COVID-19 na Índia')

# Gráfico obtido
plt.show()
```

```python
# Gráfico de dispersão

# Casos acumulados
plt.scatter(df_t['Date'], df_t['Total Confirmed'], label='Confirmados', color='y', marker='o', s=100)
plt.scatter(df_t['Date'], df_t['Total Recovered'], label='Recuperados', color='g', marker='o', s=100)
plt.scatter(df_t['Date'], df_t['Total Deceased'], label='Mortos', color='r', marker='+', s=100)
plt.legend()
plt.xticks([])
plt.xlabel('Tempo em Dias (30/01 a 13/05)')
plt.ylabel('Número de Casos Acumulados')
plt.title('Casos Acumulados de COVID-19 na Índia')

# Gráfico obtido
plt.show()
```
