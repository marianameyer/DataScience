```python
# Gráfico em linha do consumo em todo o período
df['Consumption'].plot(linewidth=0.2, color='red', grid='on', figsize=(10,6))
plt.xlabel('Tempo [anos]')
plt.ylabel('Consumo')
plt.title('Consumo de Energia', size=14);
```

```python
# Gráfico de linha do consumo apenas em 2017
df['2017']['Consumption'].plot(color='magenta', linewidth=1.2, grid='on', figsize=(10,6))
plt.xlabel('Tempo [meses]')
plt.ylabel('Consumo')
plt.title('Consumo de Energia em 2017', size=14);
```

```python
# Consumo de energia solar em 2017
df['2017']['Solar'].plot(color='red', linewidth=1.2, grid='on', figsize=(10,6))
plt.xlabel('Tempo [meses]')
plt.ylabel('Consumo')
plt.title('Consumo de Energia Solar em 2017', size=14);
```

```python
# Consumo de energia eólica em 2017
df['2017']['Wind'].plot(color='blue', linewidth=1.2, grid='on', figsize=(10,6))
plt.xlabel('Tempo [meses]')
plt.ylabel('Consumo')
plt.title('Consumo de Energia Eólica em 2017', size=14);
```

Consumo Solar e Eólico Semanal:

```python
# Reamostragem dos dados para frequência semanal em relação à média
df_week = df.resample('W').mean()
```

```python
# Comparação lado a lado
fig = plt.figure(figsize=(18, 10))

ax1 = fig.add_subplot(1, 2, 1)
ax1.plot(df_week['2006':].index, df_week.loc['2006':, 'Solar'], color='orange', linewidth=1)
ax1.set_title('Solar')
ax1.set_ylabel('Consumo')
ax1.tick_params(axis='x', labelrotation = 45)
ax1.grid(True)

ax2 = fig.add_subplot(1, 2, 2)
ax2.plot(df_week['2006':].index, df_week.loc['2006':, 'Wind'], color='blue', linewidth=1)
ax2.set_title('Eólica')
ax2.tick_params(axis='x', labelrotation = 45)
ax2.grid(True)

plt.suptitle('Consumo Semanal', size=16)
plt.show()
```

```python
# Comparação no mesmo gráfico
plt.figure(figsize=(14, 6))

plt.plot(df_week['2006':].index, df_week.loc['2006':, 'Wind'], color='blue', linewidth=1.2, label="Eólica")
plt.plot(df_week['2006':].index, df_week.loc['2006':, 'Solar'], color='orange', linewidth=1.2, label="Solar")

plt.legend()
plt.xlabel('Tempo')
plt.ylabel('Consumo')
plt.title('Comparativo Consumo Semanal', size=14)
plt.grid(True)

plt.show()
```

Consumo Solar e Eólico Diário:

```python
# Reamostragem dos dados para frequência diária em relação à média
df_daily = df.resample('D').mean()
```

```python
# # Comparação lado a lado
fig = plt.figure(figsize=(18, 10))

ax1 = fig.add_subplot(1, 2, 1)
ax1.plot(df_daily['2006':].index, df_daily.loc['2006':, 'Solar'], color='orange', linewidth=0.5)
ax1.set_title('Solar')
ax1.set_ylabel('Consumo')
ax1.tick_params(axis='x', labelrotation = 45)
ax1.grid(True)

ax2 = fig.add_subplot(1, 2, 2)
ax2.plot(df_daily['2006':].index, df_daily.loc['2006':, 'Wind'], color='blue', linewidth=0.5)
ax2.set_title('Eólica')
ax2.tick_params(axis='x', labelrotation = 45)
ax2.grid(True)

plt.suptitle('Consumo Diário', size=16)
plt.show()
```

```python
# Comparação no mesmo gráfico
plt.figure(figsize=(14, 6))

plt.plot(df_daily['2006':].index, df_daily.loc['2006':, 'Wind'], color='blue', linewidth=1.2, label="Eólica")
plt.plot(df_daily['2006':].index, df_daily.loc['2006':, 'Solar'], color='orange', linewidth=1.2, label="Solar")

plt.legend()
plt.xlabel('Tempo')
plt.ylabel('Consumo')
plt.title('Comparativo Consumo Diário', size=14)
plt.grid(True)

plt.show()
```
